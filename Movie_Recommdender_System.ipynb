```python
#genres
# id
# keywords
# title
# overview
# 'cast', 'crew'

movies = movies[['id','genres','title','keywords','overview','cast', 'crew']]
```

```python
movies.dropna(inplace = True)
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
           L.append(i['name'])
    return L
```

```python
# Used to convert an string format text to an list
import ast
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
def c_convert(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter += 1
        else:
            break
    return L
```

```python
movies['cast'] = movies['cast'].apply(c_convert)
```

```python
movies['cast'].iloc[0]
```

```python
movies['crew'][0]
```

```python
def cr_convert(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
```

```python
movies['crew'] = movies['crew'].apply(cr_convert)
```

```python
#used to convert a string paragraph into an list
movies['overview']= movies['overview'].apply(lambda x:x.split())
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['cast'] + movies['crew']+movies['keywords']
```

```python
movies.head()
```

```python
movies.columns
```

```python
new_df = movies[['id','title','tags']]
```

```python
#used to convert the string column to List
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
new_df.head()
```

```python
# This technique is used to correct the word in a list like:
# [Loving,love,loved] to [Love,Love,Love]
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
# To find the similarity between tags we have to do vectorization , bcz we cannt find similarities optimally , thats why we do vectorization !!
# To convert the text to vector we have used a textVectorization Method called (Beg of Words) !!
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000 , stop_words = 'english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors[0]
```

```python
vectors
```

```python
# Whenever we have worked in any larger dimension , the euclidien technique gets degrade's as dimension's increases
# So , instead we use Cosine distance , Theta between two movies for similarity

from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
# So ,when we are doing that sorting we have lose Index , which shows the similarty of Index X with Index Y,that's why we
# done list(enumerate(similarity[0])) -> so here we have done see the similarity of index 0 with any other similar Index!!

# key = lambda x:x[1] -> shows we have to sort this on the basis of 1st index
```

```python
def recommend(movie):
  movie_index = new_df[new_df['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)),reverse = True,key = lambda x:x[1])[1:6]

  for i in movies_list:
    print(new_df.iloc[i[0]].title)
    movie_id = movies.iloc[i[0]].id
    print(movie_id)
```

```python
recommend("Pirates of the Caribbean: At World's End")
```
